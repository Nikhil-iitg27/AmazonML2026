```python
for i in range(5):
  print("============================= The company from source 1: ==============================================")
  print(df1.loc[df1['entity_id'] == dft.iloc[i]['source1_entity_id']])

  print("=============== The companies matching from 2 and 3: =================")
  id_list = dft.iloc[i]['matched_entity_ids'].split(',')
  for id in id_list:
    if not df2.loc[df2['entity_id'] == id].empty:
      print(df2.loc[df2['entity_id'] == id])
    if not df3.loc[df3['entity_id'] == id].empty:
      print(df3.loc[df3['entity_id'] == id])
```
